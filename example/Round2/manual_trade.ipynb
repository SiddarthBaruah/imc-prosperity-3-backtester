```python
containers = {
    1: [80,6],
    2: [10,1],
    3: [90,10],
    4: [20,2],
    5: [73,4],
    6: [31,2],
    7: [37,3],
    8: [17,1],
    9: [50,4],
    10: [89,8]
}
```

```python
for key, values in containers.items():
    containers[key] = [values[0], values[1], 10000*values[0] / values[1], 10000*values[0] / (values[1]+10)]
```

```python
sorted_containers = dict(
    sorted(containers.items(), key=lambda item: item[1][-1], reverse=True))
sorted_containers
```

```python
total_sum = 0
for key, values in containers.items():
    total_sum += values[2]

for key, values in containers.items():
    containers[key] = [values[0], values[1], 10000 *
                       values[0] / values[1],values[2]/total_sum, 10000*values[0] / (values[1]+100*values[2]/total_sum)]
```

```python
sorted_containers = dict(sorted(containers.items(), key=lambda item: item[1][-1], reverse=True))
sorted_containers
```

```python
total_sum = 0
for key, values in containers.items():
    total_sum += values[3]

for key, values in containers.items():
    containers[key] = [values[0], values[1], 10000 *
                       values[0] / values[1], values[2]/total_sum,values[3]/total_sum, 10000*values[0] / (values[1]+100*values[3]/total_sum)]
```

```python
sorted_containers = dict(
    sorted(containers.items(), key=lambda item: item[1][-1], reverse=True))
sorted_containers
```
